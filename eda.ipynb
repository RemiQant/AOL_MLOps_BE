```python
import pandas as pd
import matplotlib.pyplot as plt

df['date'] = pd.to_datetime(df['date'])
df = df.sort_values('date')

fig, ax1 = plt.subplots(figsize=(12, 5))

ax1.plot(df['date'], df['pm2_5'], label='PM2.5',
         color='tab:blue', alpha=0.7)
ax1.set_xlabel('Date')
ax1.set_ylabel('PM2.5')
ax1.set_title('Jakarta Air Quality Trends')
ax1.grid(alpha=0.3)

ax2 = ax1.twinx()
ax2.plot(df['date'], df['us_aqi'], label='US_AQI',
         color='tab:red', alpha=0.6)
ax2.set_ylabel('US_AQI')

fig.legend(loc='upper left', bbox_to_anchor=(0.1, 0.9))
plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt

cols = [
    'pm10', 'pm2_5', 'carbon_monoxide',
    'nitrogen_dioxide', 'sulphur_dioxide',
    'ozone', 'us_aqi', 'european_aqi'
]

corr = df[cols].corr()

fig, ax = plt.subplots(figsize=(10, 7))

im = ax.imshow(corr, cmap='coolwarm', vmin=-1, vmax=1)

ax.set_xticks(range(len(cols)))
ax.set_yticks(range(len(cols)))
ax.set_xticklabels(cols, rotation=45, ha='right')
ax.set_yticklabels(cols)

for i in range(len(cols)):
    for j in range(len(cols)):
        ax.text(j, i, f'{corr.iloc[i, j]:.2f}',
                ha='center', va='center', fontsize=8)

ax.set_title('Correlation Between Air Pollutants')
fig.colorbar(im, ax=ax, label='Correlation')
plt.tight_layout()
plt.show()
```
