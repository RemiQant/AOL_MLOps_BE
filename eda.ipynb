### 14–15 Des 2022: semua polutan anjlok sekitar 10x secara bersamaan, dianggap error sumber data, diganti NaN lalu diinterpolasi

```python
df.nsmallest(10, "pm2_5")[["date", "pm2_5", "pm10", "carbon_monoxide"]]
```

```python
tanggal_rusak = pd.to_datetime(["2022-12-14", "2022-12-15"])
kolom_fitur = df.columns.drop("date")

mask = df["date"].isin(tanggal_rusak)

df.loc[mask, kolom_fitur] = np.nan

df[kolom_fitur] = df[kolom_fitur].interpolate(method="linear")
```
